```python
class GridWorld:
    def __init__(self):
        self.grid_size = 5
        self.goal = [4, 4]
        self.agent_pos = None
        self.actions = ["UP", "LEFT", "D0WN", "RIGHT"]

    def reset(self):
        self.agent_pos = [0, 0]
        return self.agent_pos
    
    def step(self, action):

        reward = 0

        if action == 0: # UP
            if self.valid_action(self.agent_pos[1], 1):
                self.agent_pos[1] += 1
                done = self.agent_pos == self.goal
                reward = 1 if done else -0.01
                return self.agent_pos, reward, done
            return self.agent_pos, -0.1, False
        
        elif action == 1: # LEFT
            if self.valid_action(self.agent_pos[0], -1):
                self.agent_pos[0] -= 1
                done = self.agent_pos == self.goal
                reward = 1 if done else -0.01
                return self.agent_pos, reward, done
            return self.agent_pos, -0.1, False
            
        elif action == 2: # DOWN
            if self.valid_action(self.agent_pos[1], -1):
                self.agent_pos[1] -= 1
                done = self.agent_pos == self.goal
                reward = 1 if done else -0.01
                return self.agent_pos, reward, done
            return self.agent_pos, -0.1, False
        
        elif action == 3: # RIGHT
            if self.valid_action(self.agent_pos[0], 1):
                self.agent_pos[0] += 1 
                done = self.agent_pos == self.goal
                reward = 1 if done else -0.01
                return self.agent_pos, reward, done
            return self.agent_pos, -0.1, False
        
        else:
            return self.agent_pos, -0.1, False

    # def step(self, action):
    #     action_map = {
    #         "UP":    (1, 1),   # (axis_index, delta)
    #         "DOWN":  (1, -1),
    #         "LEFT":  (0, -1),
    #         "RIGHT": (0, 1),
    #     }

    #     if action in action_map:
    #         axis, delta = action_map[action]
    #         if self.valid_action(self.agent_pos[axis], delta):
    #             self.agent_pos[axis] += delta

    #     done = self.agent_pos == self.goal
    #     reward = 1.0 if done else -0.01
    #     return self.agent_pos, reward, done

    def valid_action(self, state, action):
        next_step = state + action
        return next_step >= 0 and next_step < self.grid_size

    def render(self):
        for y in range(self.grid_size - 1, -1, -1):
            row = ""
            for x in range(self.grid_size):
                if [x, y] == self.agent_pos:
                    row += " A "
                elif [x, y] == self.goal:
                    row += " G "
                else:
                    row += " . "
            print(row)
        print()
                                                                                                     
```
